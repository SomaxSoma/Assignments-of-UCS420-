### Q4

```python
X = np.array([[6, 70, 3],
              [5, 50, 6],
              [8, 80, 2],
              [4, 30, 8]])
y = np.array([40, 65, 30, 85])
```

a)

```python
print(X.shape)
print(X.ndim)
```

b)

```python
print(X.T)
```

The transpose swaps rows and columns. In X each row is a user. In X.T each row is a feature (sleep, activity, stress) with the values for all 4 users.

c)

```python
xtx = X.T @ X
print(xtx)
```

d)

```python
print(np.linalg.det(xtx))
inv = np.linalg.inv(xtx)
print(inv)
```

The determinant is not 0, so X.T @ X can be inverted. X itself is 4x3, which is not square, so it cannot be inverted.

e)

```python
beta = inv @ X.T @ y
print(beta)
```

f) The coefficients tell us how much the assistance score changes when one feature increases by 1 and the others stay the same.
- sleep: about -0.16, so more sleep slightly lowers the score
- activity: about 0.14, a small effect
- stress: about 10.06, so stress has the biggest effect on the score

There are only 4 users and no intercept, so these values are not very reliable.

g)

```python
new_user = np.array([5, 40, 7])
pred = new_user @ beta
print("predicted score:", pred)
```
