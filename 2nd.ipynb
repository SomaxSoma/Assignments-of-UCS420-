## Q2: Tuple

```python
digits = [int(d) for d in "1024170194"]
L_orig = [d * 10 for d in digits]
scores = tuple(L_orig[:8])
print("scores --> ", scores)
```

```python
highest = max(scores)
print("Highest score --> ", highest, " at index --> ", scores.index(highest))

lowest = min(scores)
print("Lowest score --> ", lowest, " occurs --> ", scores.count(lowest), " times")
```

```python
reversed_scores = list(scores[::-1])
print("Reversed scores as list --> ", reversed_scores)
 
```

```python
search_score = int(input("Enter a score to search: "))
if search_score in scores:
    print("First occurrence index --> ", scores.index(search_score))
else:
    print("Score not present in tuple")
```

```python
try:
    scores[0] = 100
except TypeError as e:
    print("Error --> ", e)
 
```

```python
first_score, second_score, *remaining_scores = scores
print("First --> ", first_score)
print("Second --> ", second_score)
print("Remaining --> ", remaining_scores)
```

## Q3: Random Numbers

```python
import random
random.seed(1024170194)

nums = [random.randint(100, 900) for _ in range(100)]
print("Random Numbers --> ", nums)
```

```python
odd_count = sum(1 for x in nums if x % 2 != 0)
print("Odd numbers count --> ", odd_count)
```

```python
even_count = sum(1 for x in nums if x % 2 == 0)
print("Even numbers count --> ", even_count)
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_list = [x for x in nums if is_prime(x)]
print("Prime numbers count --> ", len(prime_list))
print("Prime numbers list --> ", prime_list)
```

```python
from collections import Counter
freq = Counter(nums)
most_common_num, most_common_count = freq.most_common(1)[0]
print("Most frequent number --> ", most_common_num, " occurs --> ", most_common_count, " times")
```

```python
print("Union A U B --> ", A.union(B))
```

```python
print("Intersection A n B --> ", A.intersection(B))
```

```python
print("A - B --> ", A.difference(B))
print("B - A --> ", B.difference(A))
 
```

```python
print("Symmetric Difference --> ", A.symmetric_difference(B))
```

```python
print("Is A subset of B --> ", A.issubset(B))
print("Is B superset of A --> ", B.issuperset(A))
```
