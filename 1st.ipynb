```python
n = int(input("Enter n: "))
total = 0
for i in range(2, n+1):
    is_prime = True
    for j in range(2, int(i**0.5) + 1):
        if i % j == 0:
            is_prime = False
            break
    if is_prime:
        total += i
print("Sum of prime numbers is", total)
```

```python
def sum_odd(n):
    total = 0
    for i in range(1, n+1):
        if i % 2 != 0:
            total += i
    return total

n = int(input("Enter n: "))
print("Sum of odd numbers is", sum_odd(n))
```

```python
def is_prime(num):
    if num < 2:
        return False
    for j in range(2, int(num**0.5) + 1):
        if num % j == 0:
            return False
    return True

def sum_primes(n):
    total = 0
    for i in range(2, n+1):
        if is_prime(i):
            total += i
    return total

n = int(input("Enter n: "))
print("Sum of prime numbers is", sum_primes(n))
```
